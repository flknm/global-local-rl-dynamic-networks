## Local files and reproducibility

`resolved_config.json` records the settings of each training run. `metrics.jsonl` contains its update logs; `summary.json` records completion, and `final_checkpoint.pt` holds the trained policy and recovery state. Saved evaluations are individual JSON files beside the checkpoint or under `runs/heuristic_evaluations/`.
